```python
def calculate_average(series):
    french_mid, french_final, french_oral, cspc_mid, cspc_final, cspc_pw, dsa, math_mid, aom_mid, aom_final, thermoexam, pointmech, frontend_final, frontendPW, frontendProject, fluid, math_particip, labs, mathf = series
    return (
        0.5 * french_mid +
        0.75 * french_final +
        0.75 * french_oral +
        0.37 * cspc_mid +
        0.73 * cspc_final +
        0.73 * cspc_pw +
        3.67 * dsa +
        1.33 * math_mid +
        0.44 * aom_mid +
        0.88 * aom_final +
        1.6 * thermoexam +
        1.6 * pointmech +
        0.92 * frontend_final +
        0.18 * frontendPW +
        0.73 * frontendProject +
        0.8 * fluid +
        0.67 * math_particip +
        1.33 * labs +
        2 * mathf
    ) / (0.5 + 0.75 + 0.75 + 0.37 + 0.73 + 0.73 + 3.67 + 1.33 + 0.44 + 0.88 + 1.6 + 1.6 + 0.92 + 0.18 + 0.73 + 0.8 + 0.67 + 1.33 + 2)
```
